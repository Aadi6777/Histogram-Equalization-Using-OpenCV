### Write a Python program using OpenCV to perform histogram equalization on the given grayscale image **"parrot.jpg"**.

The program should:

- Import the required libraries (**OpenCV, NumPy, Matplotlib**).
- Read the image **"parrot.jpg"** in **grayscale** format.
- Display the grayscale image and plot its histogram.
- Perform **histogram equalization** on the grayscale image using `cv2.equalizeHist()` to enhance its contrast.
- Using Matplotlib, display the following in a **2 × 2 grid layout**:
  - Original Grayscale Image
  - Histogram of Original Image
  - Enhanced (Equalized) Image
  - Histogram of Enhanced Image


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

```

```python
# Read the image in grayscale format
img_bgr= cv2.imread("parrot.jpg",0)
```

```python
# Display the grayscale image.
plt.imshow(img_bgr,cmap="grey")
plt.title("go")

```

```python
# Plot the histogram of the grayscale image
plt.hist(img_bgr.ravel(),256,range=[0,256])
plt.title("Histogram of Original Image")
plt.show()

```

```python
# Perform histogram equalization
equalized_img = cv2.equalizeHist(img_bgr)

```

```python
# Display [1] the Original Image (Gray Image) and its Histogram, and [2] the Enhanced Image and its Histogram using a 2×2 layout in Matplotlib.

# Display the Original Image and Histogram
# Display the Enhanced Image and Histogram

plt.figure(figsize=(12,10))

# Original Grayscale Image
plt.subplot(2,2,1)
plt.imshow(img_bgr, cmap='gray')
plt.title("Original Grayscale Image")


# Histogram of Original Image
plt.subplot(2,2,2)
plt.hist(img_bgr.ravel(), bins=256, range=[0,256])
plt.title("Histogram of Original Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

# Equalized Image
plt.subplot(2,2,3)
plt.imshow(equalized_img, cmap='gray')
plt.title("Equalized Image")
plt.axis('off')

# Histogram of Equalized Image
plt.subplot(2,2,4)
plt.hist(equalized_img.ravel(), bins=256, range=[0,256])
plt.title("Histogram of Equalized Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


```
